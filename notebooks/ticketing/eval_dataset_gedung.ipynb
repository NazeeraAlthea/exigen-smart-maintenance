```python
# ==============================================================
# SCRIPT PEMBERSIH TYPO KATA GANDA (Misal: "Gedung Gedung B")
# ==============================================================
import pandas as pd
import re
import os

# 1. SETUP PATH FILE (Sesuaikan dengan struktur folder Anda)
path_kotor = "../../data/dataset_tiket_lengkap.csv"
path_bersih = "../../data/dataset_tiket_lengkap_gedung.csv"

print(f"📥 Membaca data kotor dari: {path_kotor}...")

try:
    # Load dataset
    df = pd.read_csv(path_kotor, sep='|', on_bad_lines='skip')
    print(f"📊 Total data yang akan dibersihkan: {len(df)} baris\n")

    # 2. FUNGSI REGEX PEMBASMI KATA GANDA
    def basmi_gedung_ganda(text):
        if pd.isna(text):
            return text
        text_str = str(text)
        
        # Jaga-jaga jika AI sampai halusinasi menulis "Gedung Gedung Gedung" (3x atau lebih)
        text_str = re.sub(r'(?i)\b(gedung\s+){2,}', 'Gedung ', text_str)
        
        # Regex ini mencari kata "gedung gedung" (bisa huruf besar/kecil, atau singkatan gd/gdng)
        # dan merubahnya menjadi "Gedung " saja.
        text_str = re.sub(r'(?i)\b(gedung|gdng|gd\.?)\s+(gedung|gdng|gd\.?)\b', 'Gedung', text_str)
        
        return text_str

    # 3. TERAPKAN KE KOLOM YANG TERDAMPAK
    print("🧹 Menyapu bersih duplikasi kata 'Gedung Gedung'...")
    kolom_sasaran = ['teks_keluhan_awam', 'teks_laporan_teknisi', 'lokasi_gedung']
    
    for col in kolom_sasaran:
        if col in df.columns:
            df[col] = df[col].apply(basmi_gedung_ganda)
            
    # Tampilkan pembuktian (sebelum disimpan)
    print("\n✨ Pembuktian! Cuplikan Teks Setelah Dibersihkan:")
    for teks in df['teks_keluhan_awam'].head(3):
        print(f"-> {teks}")

    # 4. SIMPAN KE FILE BARU
    df.to_csv(path_bersih, sep='|', index=False)
    print(f"\n💾 BERHASIL! File super bersih telah disimpan di: {path_bersih}")

except FileNotFoundError:
    print(f"❌ Error: File {path_kotor} tidak ditemukan. Pastikan Anda berada di direktori yang tepat!")
except Exception as e:
    print(f"❌ Terjadi kesalahan: {e}")
```
